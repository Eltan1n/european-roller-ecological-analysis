The headline change since 2024 is whole-CNS and sex-specific connectomes. Let's list what's live right now rather than trusting a hard-coded list.

````python
# All datasets VFB integrates (verified API method) — returns ~200 rows incl. LM & scRNAseq.
datasets = vfb.get_datasets()
print(len(datasets), 'datasets total')
datasets[['label', 'symbol', 'id']].head(20)
# Tip: the no-code MCP tool's list_connectome_datasets returns just the connectomes.
````

````python
from vfb_connect import vfb
import pandas as pd

hedef_hucre = "MBON-gamma5" # Veya literatürdeki alternatif adıyla MBON-01

print(f"1. {hedef_hucre} hücresine sinyal gönderen (upstream) hücreler aranıyor...")
# Bu komut, hedef hücreye bağlanan tüm nöronları ve sinaps sayılarını çeker
girdiler = vfb.get_neurons_upstream_of(hedef_hucre, return_dataframe=True)

if girdiler is not None and not girdiler.empty:
    print(f"\nBAŞARILI! {hedef_hucre} hücresine bağlanan toplam {len(girdiler)} farklı nöron grubu bulundu.")

    print("\n2. Sinaps sayısına (Weight) göre en güçlü 5 partner sıralanıyor:")
    # İlgili sütunları seçip sinaps ağırlığına (weight) göre büyükten küçüğe sıralıyoruz
    # Sütun isimleri versiyona göre değişebildiği için en temel olanları alıyoruz
    gosterilecek_sutunlar = [col for col in girdiler.columns if col in ['id', 'label', 'weight', 'symbol']]

    en_gucluler = girdiler[gosterilecek_sutunlar].sort_values(by='weight', ascending=False)
    print(en_gucluler.head(10))

    print("\n(Dikkat et, bu listedeki en güçlü partnerlerin 'Kenyon Cell' veya 'KC' olması gerekiyor!)")
else:
    print(f"HATA: VFB, {hedef_hucre} için veri döndürmedi. İsimlendirme farklı olabilir (Örn: 'MBON01').")
````
